# Tugas: Singular Value Decomposition (SVD)

## Aplikasi SVD pada Kompresi Gambar Berwarna

Di bawah ini adalah kode untuk mengompresi gambar menggunakan metode SVD. Perhatikan penggunaan path berkas gambar yang tepat dan pemisahan channel warna (R, G, B).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Membaca gambar (OpenCV membaca dalam format BGR)
color_img = cv2.imread('../images/foto_kucing.png')

if color_img is None:
    raise FileNotFoundError("Gambar '../images/foto_kucing.png' tidak ditemukan. Pastikan file ada di folder images/.")

# Memisahkan channel warna B, G, R
B, G, R = cv2.split(color_img)

# Fungsi untuk melakukan kompresi pada masing-masing channel
def kompresi_rgb(R, G, B, n):
    hasil = []
    for channel in [R, G, B]:
        # Hitung SVD pada satu channel warna
        U, S, Vt = np.linalg.svd(channel, full_matrices=False)
        
        # Rekonstruksi matriks menggunakan n nilai singular pertama
        compressed = U[:, :n] @ np.diag(S[:n]) @ Vt[:n, :]
        
        # Memastikan nilai piksel berada dalam batas [0, 255]
        hasil.append(np.clip(compressed, 0, 255))
        
    # Gabungkan kembali channel warna terkompresi (menjadi format RGB untuk matplotlib)
    return cv2.merge(hasil).astype(np.uint8)

# Tampilkan gambar asli
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(color_img, cv2.COLOR_BGR2RGB))
plt.title('Gambar Asli')
plt.axis('off')

# Tampilkan gambar terkompresi dengan n=50
plt.subplot(1, 2, 2)
plt.imshow(kompresi_rgb(R, G, B, n=50))
plt.title('Gambar Terkompresi (n=50)')
plt.axis('off')

plt.show()
```
