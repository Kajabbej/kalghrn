# Tugas: Singular Value Decomposition (SVD)

## Bagian 1: Teori dan Perhitungan Matriks

Diberikan matriks $A$ sebagai berikut:
$$
A = \begin{bmatrix} 3 & 2 & 2 \\ 2 & 3 & -2 \end{bmatrix}
$$

1. Carilah nilai singular dari matriks $A$.
2. Carilah matriks $U$, $\Sigma$, dan $V^T$ sedemikian rupa sehingga $A = U \Sigma V^T$.
3. Verifikasi hasil pemfaktoran Anda dengan mengalikan kembali matriks-matriks tersebut.

```python
# Tuliskan kode penyelesaian Bagian 1 Anda di sini
import numpy as np

# Definisikan matriks A
A = np.array([
    [3, 2, 2],
    [2, 3, -2]
])

# Hitung SVD
U, S, Vt = np.linalg.svd(A)
print("Matriks U:\n", U)
print("Nilai Singular (S):\n", S)
print("Matriks V^T:\n", Vt)
```

## Bagian 2: Aplikasi SVD pada Kompresi Gambar Berwarna

Di bawah ini adalah kode untuk mengompresi gambar profil menggunakan metode SVD. Perhatikan penggunaan path berkas gambar yang tepat dan pemisahan channel warna (R, G, B).

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Membaca gambar (OpenCV membaca dalam format BGR)
color_img = cv2.imread('../images/myprofile.jpg')

if color_img is None:
    raise FileNotFoundError("Gambar '../images/myprofile.jpg' tidak ditemukan. Pastikan file ada di folder images/.")

# Memisahkan channel warna B, G, R
B, G, R = cv2.split(color_img)

# Fungsi untuk melakukan kompresi pada masing-masing channel
def kompresi_rgb(R, G, B, n):
    hasil = []
    for channel in [R, G, B]:
        # Hitung SVD pada satu channel warna
        U, S, Vt = np.linalg.svd(channel, full_matrices=False)
        
        # Rekonstruksi matriks menggunakan n nilai singular pertama
        compressed = U[:, :n] @ np.diag(S[:n]) @ Vt[:n, :]
        
        # Memastikan nilai piksel berada dalam batas [0, 255]
        hasil.append(np.clip(compressed, 0, 255))
        
    # Gabungkan kembali channel warna terkompresi (menjadi format RGB untuk matplotlib)
    return cv2.merge(hasil).astype(np.uint8)

# Tampilkan gambar asli
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(color_img, cv2.COLOR_BGR2RGB))
plt.title('Gambar Asli')
plt.axis('off')

# Tampilkan gambar terkompresi dengan n=50
plt.subplot(1, 2, 2)
plt.imshow(kompresi_rgb(R, G, B, n=50))
plt.title('Gambar Terkompresi (n=50)')
plt.axis('off')

plt.show()
```
