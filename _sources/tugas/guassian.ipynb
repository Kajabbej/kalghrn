# tugas gauss

Diketahui:

$$
\begin{cases}
x + y + z + w = 10 \\
2x - y + z + w = 5 \\
3x + y - z + w = 12 \\
x + 2y + 3z - w = 7
\end{cases}
$$

---

### Langkah 1: Matriks Augmentasi

$$
\left[
\begin{array}{cccc|c}
1 & 1 & 1 & 1 & 10 \\
2 & -1 & 1 & 1 & 5 \\
3 & 1 & -1 & 1 & 12 \\
1 & 2 & 3 & -1 & 7
\end{array}
\right]
$$

---

### Langkah 2: Eliminasi Kolom Pertama

Lakukan operasi baris:

$$
R_2 = R_2 - 2R_1
$$

$$
R_3 = R_3 - 3R_1
$$

$$
R_4 = R_4 - R_1
$$

Hasil:

$$
\left[
\begin{array}{cccc|c}
1 & 1 & 1 & 1 & 10 \\
0 & -3 & -1 & -1 & -15 \\
0 & -2 & -4 & -2 & -18 \\
0 & 1 & 2 & -2 & -3
\end{array}
\right]
$$

---

### Langkah 3: Eliminasi Kolom Kedua

Gunakan pivot pada baris kedua.

$$
R_3 = R_3 - \frac{2}{3}R_2
$$

$$
R_4 = R_4 + \frac{1}{3}R_2
$$

Hasil:

$$
\left[
\begin{array}{cccc|c}
1 & 1 & 1 & 1 & 10 \\
0 & -3 & -1 & -1 & -15 \\
0 & 0 & -\frac{10}{3} & -\frac{4}{3} & -8 \\
0 & 0 & \frac{5}{3} & -\frac{7}{3} & -8
\end{array}
\right]
$$

---

### Langkah 4: Eliminasi Kolom Ketiga

Gunakan pivot pada baris ketiga.

$$
R_4 = R_4 + \frac{1}{2}R_3
$$

Hasil:

$$
\left[
\begin{array}{cccc|c}
1 & 1 & 1 & 1 & 10 \\
0 & -3 & -1 & -1 & -15 \\
0 & 0 & -\frac{10}{3} & -\frac{4}{3} & -8 \\
0 & 0 & 0 & -3 & -12
\end{array}
\right]
$$

Matriks telah berbentuk eselon baris.

---

### Langkah 5: Substitusi Balik

#### Baris 4

$$
-3w=-12
$$

$$
w=4
$$

---

#### Baris 3

$$
-\frac{10}{3}z-\frac{4}{3}(4)=-8
$$

$$
-\frac{10}{3}z-\frac{16}{3}=-8
$$

Kalikan dengan 3:

$$
-10z-16=-24
$$

$$
-10z=-8
$$

$$
z=\frac{4}{5}
$$

---

#### Baris 2

$$
-3y-z-w=-15
$$

$$
-3y-\frac{4}{5}-4=-15
$$

$$
-3y=-15+\frac{24}{5}
$$

$$
-3y=-\frac{51}{5}
$$

$$
y=\frac{17}{5}
$$

---

#### Baris 1

$$
x+y+z+w=10
$$

$$
x+\frac{17}{5}+\frac{4}{5}+4=10
$$

$$
x+\frac{21}{5}+4=10
$$

$$
x+\frac{41}{5}=10
$$

$$
x=\frac{50}{5}-\frac{41}{5}
$$

$$
x=\frac{9}{5}
$$

---

### Solusi SPL 4 Variabel

$$
x=\frac{9}{5}, \qquad
y=\frac{17}{5}, \qquad
z=\frac{4}{5}, \qquad
w=4
$$

atau

$$
(x,y,z,w)=
\left(
\frac{9}{5},
\frac{17}{5},
\frac{4}{5},
4
\right)
$$
