# 📋 Konsep Dasar Matriks

Matriks adalah susunan bilangan, simbol, atau ekspresi berbentuk persegi panjang yang diatur dalam baris dan kolom. Di dalam notebook ini, kita akan mempelajari representasi matriks dan operasi dasar matriks menggunakan pustaka **NumPy**.

## 1. Mendefinisikan Matriks

Matriks diidentifikasi berdasarkan ordonya (jumlah baris $\times$ jumlah kolom). Berikut adalah contoh mendefinisikan matriks $A$ berordo $2 \times 2$ dan matriks $B$ berordo $2 \times 2$ di Python:

```python
import numpy as np

A = np.array([[4, 7],
              [2, 6]])

B = np.array([[1, 2],
              [3, 5]])

print("Matriks A:\n", A)
print("\nMatriks B:\n", B)
```

## 2. Operasi Aritmatika Matriks

### Penjumlahan & Pengurangan
Penjumlahan/pengurangan matriks dilakukan dengan menjumlahkan/mengurangi elemen-elemen yang seletak. Kedua matriks harus memiliki ordo yang sama.

### Perkalian Matriks (Dot Product)
Perkalian matriks $A \times B$ hanya dapat dilakukan jika jumlah kolom $A$ sama dengan jumlah baris $B$.

```python
# Penjumlahan
C_tambah = A + B
print("Penjumlahan A + B:\n", C_tambah)

# Pengurangan
C_kurang = A - B
print("\nPengurangan A - B:\n", C_kurang)

# Perkalian (Dot Product)
C_kali = A @ B  # atau np.dot(A, B)
print("\nPerkalian A x B:\n", C_kali)
```

## 3. Transpos & Determinan

Transpos matriks ($A^T$) diperoleh dengan menukar baris menjadi kolom. Determinan adalah nilai numerik khusus yang hanya dapat dihitung pada matriks persegi.

```python
print("Transpos A:\n", A.T)
print("\nDeterminan A:", np.linalg.det(A))
```
