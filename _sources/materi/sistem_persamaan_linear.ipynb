# ⚙️ Sistem Persamaan Linear (SPL) & Eliminasi Gaussian

Notebook ini membahas bagaimana memodelkan **Sistem Persamaan Linear (SPL)** ke dalam bentuk persamaan matriks, serta menyelesaikannya menggunakan **Eliminasi Gaussian** langkah demi langkah dengan Python.

## 1. Representasi SPL ke Matriks

SPL dengan $m$ persamaan dan $n$ variabel dapat ditulis dalam bentuk:
$$Ax = b$$

Misal kita memiliki SPL:
$$
\begin{aligned}
2x + y + z &= 8 \\
3x - y - z &= 2 \\
x + 2y - z &= 2
\end{aligned}
$$

Mari representasikan SPL ini ke dalam **Matriks Koefisien** $A$, **Vektor Konstanta** $b$, dan **Matriks Diperbesar (Augmented Matrix)** $[A | b]$.

```python
import numpy as np

A = np.array([[2.0, 1.0, 1.0],
              [3.0, -1.0, -1.0],
              [1.0, 2.0, -1.0]])

b = np.array([8.0, 2.0, 2.0])

# Augmented Matrix
augmented = np.column_stack((A, b))
print("Matriks Koefisien A:\n", A)
print("\nVektor Hasil b:\n", b)
print("\nMatriks Diperbesar [A | b]:\n", augmented)
```

## 2. Eliminasi Gaussian (OBE)

Eliminasi Gaussian adalah algoritma untuk mengubah matriks diperbesar menjadi bentuk **Eselon Baris**. Setelah itu, nilai variabel ditentukan dengan **Substitusi Balik**.

Berikut adalah fungsi Python untuk melakukan Eliminasi Gaussian:

```python
def gaussian_elimination(A, b):
    M = np.column_stack((A, b)).astype(float)
    n = len(b)
    
    # 1. Eliminasi Maju
    for i in range(n):
        # Pencarian Pivot Maksimum (Partial Pivoting)
        max_row = np.argmax(np.abs(M[i:, i])) + i
        if M[max_row, i] == 0:
            raise ValueError("SPL tidak memiliki solusi unik.")
        M[[i, max_row]] = M[[max_row, i]]
        
        # Operasi Baris untuk membuat elemen di bawah pivot menjadi nol
        for j in range(i + 1, n):
            factor = M[j, i] / M[i, i]
            M[j, i:] -= factor * M[i, i:]
            
    # 2. Substitusi Balik
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (M[i, -1] - np.dot(M[i, i+1:n], x[i+1:n])) / M[i, i]
        
    return x

solusi = gaussian_elimination(A, b)
print("Solusi SPL hasil Eliminasi Gaussian:", solusi)
```

## 3. Penyelesaian Instan dengan NumPy

Untuk kebutuhan praktis, kita dapat memanggil `np.linalg.solve(A, b)` secara langsung.

```python
solusi_numpy = np.linalg.solve(A, b)
print("Solusi np.linalg.solve:", solusi_numpy)
```
