## 1. Lembar Uji Coba Matriks

Silakan jalankan blok kode di bawah ini untuk mencoba mendefinisikan matriks Anda sendiri. Ubah nilai di dalam `np.array` sesuai keinginan.

```python
import numpy as np

# Masukkan matriks Anda di sini
X = np.array([[1, 2],
              [3, 4]])

print("Matriks X:\n", X)
print("\nTranspos Matriks X:\n", X.T)
print("\nDeterminan Matriks X:", np.linalg.det(X))
```

## 2. Menyelesaikan SPL Anda Sendiri

Ubah matriks koefisien `A_coba` dan vektor hasil `b_coba` untuk menemukan nilai variabelnya secara otomatis.

```python
# Definisi SPL Ax = b
A_coba = np.array([[1, 1],
                   [2, -1]])

b_coba = np.array([5, 1])

try:
    solusi = np.linalg.solve(A_coba, b_coba)
    print("Solusi (x, y):", solusi)
except np.linalg.LinAlgError:
    print("Matriks koefisien tidak memiliki invers (solusi tidak tunggal/tidak ada).")
```
