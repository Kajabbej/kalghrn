# Playground: Eliminasi Gaussian

Di sini kita dapat bereksperimen dengan kode Python menggunakan NumPy maupun SageMath untuk menyelesaikan Sistem Persamaan Linear (SPL) dengan metode Eliminasi Gaussian dan substitusi balik.

## 1. Penyelesaian SPL 3 Variabel

Sistem persamaan linear:
$$
\begin{aligned}
2x + y - z &= 8 \\
-3x - y + 2z &= -11 \\
-2x + y + 2z &= -3
\end{aligned}
$$

```python
import numpy as np

# Matriks augmented [A|b]
A = np.array([
    [ 2.0,  1.0, -1.0,  8.0],
    [-3.0, -1.0,  2.0, -11.0],
    [-2.0,  1.0,  2.0, -3.0]
])

n = 3

# Forward Elimination dengan Partial Pivoting
for col in range(n):
    max_row = np.argmax(abs(A[col:, col])) + col
    A[[col, max_row]] = A[[max_row, col]]

    for row in range(col + 1, n):
        faktor = A[row, col] / A[col, col]
        A[row, col:] -= faktor * A[col, col:]

# Back Substitution
x = np.zeros(n)
for i in range(n - 1, -1, -1):
    x[i] = A[i, n]
    for j in range(i + 1, n):
        x[i] -= A[i, j] * x[j]
    x[i] /= A[i, i]

print(f"Hasil SPL 3 Variabel:")
print(f"x = {x[0]:.4f}")
print(f"y = {x[1]:.4f}")
print(f"z = {x[2]:.4f}")
```

## 2. Penyelesaian SPL 4 Variabel

Sistem persamaan linear:
$$
\begin{aligned}
w + x + y + z &= 10 \\
2w + 3x + y - z &= 14 \\
w - x + 3y + 2z &= 9 \\
3w + x - y + 2z &= 13
\end{aligned}
$$

```python
import numpy as np

# Matriks augmented [A|b] - 4 variabel
A = np.array([
    [1.0,  1.0,  1.0,  1.0,  10.0],
    [2.0,  3.0,  1.0, -1.0,  14.0],
    [1.0, -1.0,  3.0,  2.0,   9.0],
    [3.0,  1.0, -1.0,  2.0,  13.0]
])

n = 4

# Forward Elimination dengan Partial Pivoting
for col in range(n):
    max_row = np.argmax(abs(A[col:, col])) + col
    A[[col, max_row]] = A[[max_row, col]]

    for row in range(col + 1, n):
        faktor = A[row, col] / A[col, col]
        A[row, col:] -= faktor * A[col, col:]

# Back Substitution
x = np.zeros(n)
for i in range(n - 1, -1, -1):
    x[i] = A[i, n]
    for j in range(i + 1, n):
        x[i] -= A[i, j] * x[j]
    x[i] /= A[i, i]

print(f"Hasil SPL 4 Variabel:")
var_names = ['w', 'x', 'y', 'z']
for nama, val in zip(var_names, x):
    print(f"{nama} = {val:.4f}")
```
